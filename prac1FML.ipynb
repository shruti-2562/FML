```python
import pandas as pd
data = {
    "Student":["Amit","Priya","Rahul","Sakshi"],
    "Gender":["Male","Female","Male","Female"],
    "Attendance":[80,60,95,None],
    "Marks":[50,75,None,98]
}

df = pd.DataFrame(data)

print("Original Dataset")
print(df)
print(df.isnull().sum())
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["Gender"]=le.fit_transform(df["Gender"])

print(df)
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[["Attendance","Marks"]] = scaler.fit_transform(
    df[["Attendance","Marks"]]
)

print(df)
print("Average Marks =", df["Marks"].mean())
print("Highest Attendance =", df["Attendance"].max())
print(df[["Attendance","Marks"]].corr())
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import MinMaxScaler

#Load Dataset
df = pd.read_csv("student_dataset.csv")

print("Original Dataset")
print(df)

#Check Missing Values
print("\nMissing Values")
print(df.isnull().sum())

#Data Cleaning
# Corrected select_datatypes to select_dtypes and include argument
for column in df.select_dtypes(include=['float64', 'int64']).columns:
  df[column] = df[column].fillna(df[column].mean())

# Corrected select_datatype to select_dtypes and include argument
for column in df.select_dtypes(include=['object']).columns:
  df[column] = df[column].fillna(df[column].mode()[0])

print("\nDataset After Data Cleaning")
print(df)

#Encoding
le = LabelEncoder()

# Corrected indentation for print and subsequent steps
for column in df.select_dtypes(include=['object']).columns:
  df[column] = le.fit_transform(df[column])

print("\nDataset After Encoding")
print(df)

#Normalization
scaler = MinMaxScaler()
numeric_columns = df.select_dtypes(include=['int64','float64']).columns

# Corrected typo fit_tranform to fit_transform
df[numeric_columns] = scaler.fit_transform(df[numeric_columns])

print("\nDataset After Normalization")
print(df)

#Pattern Identification
print("\nDataset Statics")
print(df.describe())

print("\nCorrelation Matrix")
print(df.corr())
```
